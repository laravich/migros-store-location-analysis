# creating a complete master dataset

```python
# load all datas
import pandas as pd
import numpy as np

stores_df = pd.read_csv(
    "data/switzerland_supermarkets_clean.csv",
    dtype={"postal_code": "string"}
)

population_df = pd.read_csv(
    "data/switzerland_population_clean.csv",
    dtype={"postal_code": "string"}
)

taxpayers_df = pd.read_csv(
    "data/num_taxpayers_by_gemeinde.csv"
)

income_df = pd.read_csv(
    "data/total_income_by_gemeinde.csv"
)

postcode_municipality_df = pd.read_csv(
    "data/postal_code_municipality_mapping.csv",
    dtype={"postal_code": "string"}
)

postcode_coordinates_df = pd.read_csv(
    "data/postcode_coordinates.csv",
    dtype={"postal_code": "string"}
)
```

```python
# Display all columns without truncation
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

dataframes = {
    "stores_df": stores_df,
    "population_df": population_df,
    "taxpayers_df": taxpayers_df,
    "income_df": income_df,
    "postcode_municipality_df": postcode_municipality_df,
    "postcode_coordinates_df": postcode_coordinates_df
}

for name, dataframe in dataframes.items():
    print(f"\n{name}")
    print("-" * len(name))
    print(dataframe.columns.tolist())
```

```python
for name, dataframe in dataframes.items():
    print(f"\n{name}: shape = {dataframe.shape}")
    display(dataframe.head())
```

# cleaning Postal codes

```python
#formatting postalcode in all dataframe - so all postcodes have the same 4 digit text format
for dataframe in [stores_df, population_df, postcode_municipality_df, postcode_coordinates_df]:
    dataframe["postal_code"] = (
        dataframe["postal_code"]
        .astype("string")   # converts all into string
        .str.extract(r"(\d+)")[0] # extracts continuous digits and removes '.' or other alphabets
        .str.zfill(4) # adds 0 to create a clean 4 digits
    )
```

```python
# Find valid postcodes that exist in the population dataset
valid_population_postcodes = set(population_df["postal_code"])

valid_postcode_coordinates = (postcode_coordinates_df[postcode_coordinates_df["postal_code"].isin(valid_population_postcodes)]
    .dropna(subset=["postcode_latitude","postcode_longitude"])
    .copy()
)
```

```python
# Find stores with a missing or invalid postcode
needs_postcode_assignment = (stores_df["postal_code"].isna() | ~stores_df["postal_code"].isin(valid_population_postcodes))

print("Stores requiring postcode assignment:", needs_postcode_assignment.sum())
```

```python
stores_df["original_postal_code"] = (stores_df["postal_code"])

stores_df["postcode_source"] = np.where(
    needs_postcode_assignment,
    "Nearest valid population postcode",
    "OSM address"
)
```

```python
stores_df["postal_code"].isna().sum() #stores missing postal codes
```

```python
# Assign missing store postcodes
# Use the nearest official postcode coordinate
from sklearn.neighbors import BallTree

#known_postcodes = postcode_coordinates_df.dropna(subset=["postcode_latitude", "postcode_longitude"]).copy()

# BallTree with haversine distance requires radians
postcode_coordinates_rad = np.radians(valid_postcode_coordinates[["postcode_latitude", "postcode_longitude"]].to_numpy())

tree = BallTree(postcode_coordinates_rad, metric="haversine")

# missing_postcode_mask = stores_df["postal_code"].isna()
#changing this because it missed few stores


store_coordinates_rad = np.radians(stores_df.loc[needs_postcode_assignment, ["latitude", "longitude"]].to_numpy())

distance, index = tree.query(store_coordinates_rad, k=1) # find closest distance
#print(distance)
stores_df.loc[needs_postcode_assignment, "postal_code"] = valid_postcode_coordinates.iloc[index.flatten()]["postal_code"].to_numpy()

# Convert angular distance into kilometres
stores_df.loc[needs_postcode_assignment, "postcode_distance_km"] = distance.flatten() * 6371 #radians to KM (angular dist x earth radius in km)
```

```python

print("Stores still outside population postcodes:",( ~stores_df["postal_code"].isin(population_df["postal_code"])).sum())

print("\nStore totals after postcode assignment:")
print(stores_df["store_group"].value_counts().sort_index())
```

```python
# Mark whether a postcode came from OSM or was estimated.
# Adding this before replacing the missing values:
stores_df["postcode_source"] = np.where(stores_df["postal_code"].notna(),"OSM address", "Nearest postcode coordinate")

#checking
#print(stores_df["postcode_source"].value_counts())

#print(stores_df["postcode_distance_km"].describe())

#print("Remaining missing postcodes:",stores_df["postal_code"].isna().sum())
```

```python
# Review stores assigned unusually far away
stores_df[stores_df["postcode_distance_km"] > 4][  #Nearest coordinates are an approximation
    [
        "name",
        "store_group",
        "latitude",
        "longitude",
        "postal_code",
        "postcode_distance_km"
    ]
].sort_values("postcode_distance_km", ascending=False).head(20)
```

```python
# Combine income and taxpayers
```

```python
municipality_income_df = taxpayers_df[
    [
        "Gemeinde ID",
        "Kanton / Canton",
        "Gemeinde / Commune",
        "num_taxpayers"
    ]].merge(income_df[
        [
            "Gemeinde ID",
            "Total_income"
        ]
    ],
    on="Gemeinde ID",
    how="inner",
    validate="one_to_one"
)
```

```python
# Calculate average income per taxpayer
municipality_income_df["avg_income_per_taxpayer"] = (municipality_income_df["Total_income"] / municipality_income_df["num_taxpayers"])

municipality_income_df.replace([np.inf, -np.inf], np.nan,inplace=True)
```

```python
municipality_income_df[["num_taxpayers", "Total_income", "avg_income_per_taxpayer"]].describe()
```

```python
# Estimate postcode-level income
# Merge the municipal income with the postcode
postcode_income_df = postcode_municipality_df.merge(
    municipality_income_df[
        [
            "Gemeinde ID",
            "avg_income_per_taxpayer"
        ]
    ],
    left_on="municipality_id", # use this column on postcode_income_df
    right_on="Gemeinde ID", # use this column on postcode_municipality_df
    how="left",
    validate="many_to_one" # many postcode-mapping rows → one municipality-income row
)
postcode_income_df.head(2)
```

```python
# Calculate the weighted contribution
postcode_income_df["weighted_income"] = postcode_income_df["avg_income_per_taxpayer"]* postcode_income_df["weight_pct"] / 100
```

```python
municipality_count = (postcode_income_df
    .groupby("postal_code")["municipality_name"]
    .nunique()
    .reset_index(name="number_of_municipalities")
)

multiple_municipality_postcodes = municipality_count[
    municipality_count["number_of_municipalities"] > 1
]

multiple_municipality_postcodes
```

```python
# the same postcode can appear multiple times when it covers multiple municipalities
# Aggregate to one row per postcode
# estimating purchasing power. 
# because the crosswalk does not describe how a municipality’s total income is distributed among all its postcodes.
postcode_income_summary = (postcode_income_df.groupby("postal_code", as_index=False)
    .agg(
        avg_income_per_taxpayer=(
            "weighted_income",
            "sum"
        ),
        municipality_name=(
            "municipality_name",
            lambda x: ", ".join(sorted(set(x.dropna())))
        ),
        canton_code=(
            "canton_code",
            "first"
        )
    )
)
postcode_income_summary.head()
```

# merge stores and income based on postal codes

```python
# finding missing stores
unmatched_stores = stores_df[~stores_df["postal_code"].isin(population_df["postal_code"])].copy()

print("Number of unmatched stores:", len(unmatched_stores))

print(unmatched_stores["store_group"].value_counts())

display(
    unmatched_stores[
        [
            "name",
            "store_group",
            "postal_code",
            "city",
            "latitude",
            "longitude",
            "postcode_source",
            "postcode_distance_km"
        ]
    ]
)
```

```python
# Count stores by postcode and brand
store_counts_df = (stores_df
    .groupby(
        ["postal_code", "store_group"]
    )
    .size()
    .unstack(fill_value=0)
    .reset_index()
)
store_counts_df
```

```python
# Rename the columns
store_counts_df = store_counts_df.rename(
    columns={
        "Aldi": "aldi_count",
        "Coop": "coop_count",
        "Denner": "denner_count",
        "Lidl": "lidl_count",
        "Migros": "migros_count"
    }
)

# Ensure every expected column exists
store_count_columns = [
    "migros_count",
    "coop_count",
    "denner_count",
    "aldi_count",
    "lidl_count"
]

for column in store_count_columns:
    if column not in store_counts_df.columns:
        store_counts_df[column] = 0
```

```python
# Create the business groupings
store_counts_df["external_competitor_count"] = (
    store_counts_df["coop_count"]
    + store_counts_df["aldi_count"]
    + store_counts_df["lidl_count"]
)

store_counts_df["migros_group_count"] = (
    store_counts_df["migros_count"]
    + store_counts_df["denner_count"] # Denner is a wholly owned discount subsidiary of the Migros-Genossenschafts-Bund (MGB)
)

store_counts_df
```

# Build a master dataset

```python
population_df
```

```python
# Using the population as the base because every populated postcode should remain in the opportunity analysis
# including postcodes with zero supermarkets.
analysis_df = (
    population_df
    .merge(
        postcode_income_summary,
        on="postal_code",
        how="left",
        validate="one_to_one"
    )
    .merge(
        store_counts_df,
        on="postal_code",
        how="left",
        validate="one_to_one"
    )
    .merge(
        postcode_coordinates_df,
        on="postal_code",
        how="left",
        validate="one_to_one"
    )
)
```

```python
analysis_df.isna().sum()
```

```python
# to be safe - Fill missing store counts with zero
analysis_df[store_count_columns] = (
    analysis_df[store_count_columns]
    .fillna(0)
    .astype(int)
)

analysis_df[
    [
        "external_competitor_count",
        "migros_group_count"
    ]
] = (
    analysis_df[
        [
            "external_competitor_count",
            "migros_group_count"
        ]
    ]
    .fillna(0)
    .astype(int)
)
```

```python
analysis_df.isna().sum() 
```

```python
#six postcodes exist in the population dataset but do not have a matching row in the postcode–municipality or postcode-coordinate datasets.
missing_location_rows = analysis_df[
    analysis_df["municipality_name"].isna()
][
    [
        "postal_code",
        "population",
        "avg_income_per_taxpayer",
        "municipality_name",
        "canton_code",
        "postcode_latitude",
        "postcode_longitude",
        "locality_name"
    ]
]

missing_location_rows
```

```python
population_postcodes = set(
    population_df["postal_code"]
)

mapping_postcodes = set(
    postcode_municipality_df["postal_code"]
)

missing_postcodes = (
    population_postcodes - mapping_postcodes
)

print(missing_postcodes)
```

```python
missing_population = missing_location_rows["population"].sum()

total_population = analysis_df["population"].sum()

missing_percentage = (missing_population/ total_population * 100)

print("Population in unmatched postcodes:", missing_population)
print(
    "Percentage of Swiss population:",
    round(missing_percentage, 4),
    "%"
)
```

```python
# Create a data-quality flag for missing values
analysis_df["complete_location_data"] = (
    analysis_df[
        [
            "avg_income_per_taxpayer",
            "municipality_name",
            "canton_code",
            "postcode_latitude",
            "postcode_longitude"
        ]
    ]
    .notna()
    .all(axis=1)
)
```

```python
analysis_df["complete_location_data"].value_counts()
```

```python
# Create some useful indicators
analysis_df["has_migros"] = (analysis_df["migros_count"] > 0)

analysis_df["population_per_migros"] = (analysis_df["population"] / analysis_df["migros_count"].replace(0, np.nan))

analysis_df["population_per_migros"] = (analysis_df["population_per_migros"].fillna(analysis_df["population"]))
```

```python
import pandas as pd
import numpy as np
```

```python
#extract excel data

income_file = "data/statistik-dbst-np-gden-2022-normalfall.xlsx"

excel_file = pd.ExcelFile(income_file)

# The workbook contains one sheet for every year from 2022.
print(excel_file.sheet_names)

```

```python
# Sheet 522 is income
# Sheet 511 is number of taxpayers

# clean 511 first
taxpayers_preview = pd.read_excel(
    excel_file,
    sheet_name="511",
    header=None,
    nrows=20
)

taxpayers_preview.head(20)
```

```python
# load and skip row
taxpayers_df = pd.read_excel(
    excel_file,
    sheet_name="511",
    skiprows=2,
    header=0
)

taxpayers_df.head()

```

```python
taxpayers_df.tail(20)
```

```python
# Check null
taxpayers_df.isna().sum()
```

```python
# first fill-forword Kanton ID and Kanton/Canton

taxpayers_df["Kanton ID"] = taxpayers_df["Kanton ID"].ffill()
taxpayers_df["Kanton / Canton"] = taxpayers_df["Kanton / Canton"].ffill()

# check null value again
taxpayers_df.isna().sum()
```

```python
taxpayers_df.head()
```

```python
# convert "- to 0
convert_list = taxpayers_df.columns[4:]
convert_list
```

```python
taxpayers_clean_dash = taxpayers_df.copy()

for col in convert_list:
    taxpayers_clean_dash[col] = pd.to_numeric(taxpayers_clean_dash[col], errors='coerce').fillna(0)

taxpayers_clean_dash

```

```python
#check before convert
dash_before = (taxpayers_df[convert_list] == "- ").sum().sum()
print(f"number of dash before convert: {dash_before}")

#check after convert
dash_after =  (taxpayers_clean_dash[convert_list] == 0).sum().sum()
print(f"number of dash after convert: {dash_after}")
```

```python
# Chk duplicate ID 
taxpayers_clean_dash['Gemeinde ID'].duplicated().sum()
```

```python
taxpayers_clean_dash['Gemeinde / Commune'].value_counts()
```

```python
taxpayers_final = taxpayers_clean_dash[taxpayers_clean_dash['Gemeinde / Commune'] != 'Total / Totale']
taxpayers_final = taxpayers_final.rename(columns={"Total / Totale": "num_taxpayers"})
taxpayers_final['Gemeinde ID'].duplicated().sum()
```

```python
taxpayers_final.head()
```

# Sheet 522


```python
# load and skip row
income_df = pd.read_excel(
    excel_file,
    sheet_name="522",
    skiprows=2,
    header=0
)

income_df.head()
```

```python
# Check null
income_df.isna().sum()
```

```python
# first fill-forword Kanton ID and Kanton/Canton

income_df["Kanton ID"] = income_df["Kanton ID"].ffill()
income_df["Kanton / Canton"] = income_df["Kanton / Canton"].ffill()

# check null value again
income_df.isna().sum()
```

```python
# convert "- to 0
convert_list = income_df.columns[4:]
convert_list
```

```python
income_clean_dash = income_df.copy()

for col in convert_list:
    income_clean_dash[col] = pd.to_numeric(income_clean_dash[col], errors='coerce').fillna(0)

income_clean_dash

```

```python
#check before convert
dash_before = (income_df[convert_list] == "- ").sum().sum()
print(f"number of dash before convert: {dash_before}")

#check after convert
dash_after =  (income_clean_dash[convert_list] == 0).sum().sum()
print(f"number of dash after convert: {dash_after}")
```

```python
# Chk duplicate ID 
income_clean_dash['Gemeinde ID'].duplicated().sum()
```

```python
# Chk duplicate ID 
income_clean_dash['Gemeinde / Commune'].duplicated().sum()
```

```python
income_final = income_clean_dash[income_clean_dash['Gemeinde / Commune'] != 'Total / Totale']
income_final = income_final.rename(columns={"Total / Totale": "Total_income"})
income_final['Gemeinde ID'].duplicated().sum()
```

```python
income_final.head()
```

```python
# Export

taxpayers_final.to_csv("data/num_taxpayers_by_gemeinde.csv", index=False)
income_final.to_csv("data/total_income_by_gemeinde.csv", index=False)

print("saved:", len(taxpayers_final), "rows ->", "num_taxpayers_by_gemeinde.csv")
print("saved:", len(income_final), "rows ->", "total_income_by_gemeinde.csv")
```

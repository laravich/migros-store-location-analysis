```python
import pandas as pd
import numpy as np
import zipfile
```

```python
zip_path = "data/ortschaftenverzeichnis_plz_4326.csv.zip" 

with zipfile.ZipFile(zip_path) as z:
    print(z.namelist())
```

```python
with zipfile.ZipFile(zip_path) as z:
    with z.open("AMTOVZ_CSV_WGS84/AMTOVZ_CSV_WGS84.csv") as f:
        plz_df = pd.read_csv(f, sep=";", encoding="utf-8-sig")

plz_df.head()
```

```python
plz_df.tail()
```

```python
plz_df = plz_df.rename(columns={
    "Ortschaftsname": "locality_name",
    "PLZ4": "postal_code",
    "Zusatzziffer": "postcode_suffix",
    "Gemeindename": "municipality_name",
    "BFS-Nr": "municipality_id",
    "Kantonskürzel": "canton_code",
    "Adressenanteil": "address_share_pct",
    "Sprache": "language",
    "E": "longitude",
    "N": "latitude",
})

plz_df.head(20)
```

```python
plz_df.isna().sum()
```

```python
plz_df[plz_df.isna().any(axis=1)]
```

```python
plz_df.info()
```

```python
# convert column address_share_pct to number
plz_df["address_share_pct"] = (
    plz_df["address_share_pct"].str.replace("%", "").str.strip().astype(float)
)

plz_df["address_share_pct"].head()
```

```python
weight_df = (
    plz_df.groupby(["postal_code"])["address_share_pct"].sum()
    / plz_df.groupby("postal_code")["ZIP_ID"].nunique()
).dropna().rename("weight_pct").reset_index()

weight_df.head()
```

```python
weight_df[weight_df['weight_pct'] != 100]
```

```python
#check meaning of address_share_pct
plz_pct = plz_df.groupby(["postal_code"])["address_share_pct"].sum().reset_index()
plz_pct
```

```python
count_zip = plz_df.groupby("postal_code")["ZIP_ID"].nunique().reset_index()
count_zip
```

```python
merge_df = pd.merge(
    plz_pct,
    count_zip,
    on='postal_code',
    how="left"
)

merge_df
```

```python
# check null on merge
merge_df.isna().sum()
```

```python
# check sum = 100 in order to validate pct when analysis
merge_df['check_pct'] = merge_df['address_share_pct']/merge_df['ZIP_ID']

merge_df.head()
```

```python
# check postal_code != 100
merge_df[merge_df['check_pct'] != 100].sort_values('check_pct', ascending=False)
```

```python
merge_df[merge_df['check_pct'] != 100].sort_values('check_pct', ascending=True)
```

```python
# check each municipality (key for mappping with incometax data)
weight_df = (
    plz_df.groupby(["postal_code", "municipality_id"])["address_share_pct"].sum()
    / plz_df.groupby("postal_code")["ZIP_ID"].nunique()
).dropna().rename("weight_pct").reset_index()

weight_df.head(10)
```

```python
check = weight_df.groupby("postal_code")["weight_pct"].sum()
print((~check.between(99.9, 100.1)).sum(), "out of", len(check))
```

```python
muni_lookup = plz_df[["municipality_id", "municipality_name", "canton_code"]].drop_duplicates(subset="municipality_id")

weight_df = weight_df.merge(muni_lookup, on="municipality_id", how="left")
weight_df.head(10)
```

```python
weight_df.isna().sum()
```

```python
#export

weight_df.to_csv("data/postal_code_municipality_mapping.csv", index=False)
print("saved:", len(weight_df), "rows")
```

The weighting assumes every locality (`ZIP_ID`) within a postal code has roughly equal
  number of addresses — the source file gives no actual size/count per locality, so this is
  the best approximation available without an additional data source.
- 20 rows (postal codes 9485–9498) belong to **Liechtenstein**, not Switzerland — `canton_code`
  is missing for these since Liechtenstein has no Swiss canton. They will not match any row in
  the Swiss tax-statistics files (`num_taxpayers_by_gemeinde.csv` / `total_income_by_gemeinde.csv`).

```python
# Original plz_df also contains latitude and longitude. Export one additional postcode-coordinate file.
# Create one representative coordinate for each postal code
postcode_coordinates_df = (
    plz_df
    .groupby("postal_code", as_index=False)
    .agg(
        postcode_latitude=("latitude", "mean"),
        postcode_longitude=("longitude", "mean"),
        locality_name=("locality_name", "first")
    )
)

postcode_coordinates_df["postal_code"] = (
    postcode_coordinates_df["postal_code"]
    .astype(str)
    .str.zfill(4)
)

postcode_coordinates_df.to_csv(
    "data/postcode_coordinates.csv",
    index=False
)

print("Saved postcode coordinates:", postcode_coordinates_df.shape)
postcode_coordinates_df.head()
```
